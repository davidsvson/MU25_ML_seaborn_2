```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("banana_quality.csv")
df.head()
```

Om vi vill förutsäga Quality så har vi kategorier av värden det betyder att vi har ett klassifiseringsproblem

Om vi har en numreriskt värde med kontinuerliga värden då har vi ett regressionsproblem

```python
sns.countplot(data=df, x="Quality", hue="Quality");
```

```python
long_df = df.melt(id_vars="Quality", var_name="feature", value_name="value")
long_df.head()
```

```python
sns.displot(data=long_df, x="value", col="feature", col_wrap=4, hue="Quality", kde=True);
```

```python
sns.catplot(data=long_df, x="Quality", y="value", col="feature", hue="Quality" ,  col_wrap=4, kind="violin")
```

```python
corr = df.assign(Quality=(df["Quality"] == "Good").astype(int)).corr()

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, vmin=-1, vmax=1, cmap="coolwarm");
```

### correlation 
correlation har ett värde mellan -1 och 1. Är värdet nära -1 eller 1 har vi en stor correlation dvs samband mellan parametrarna. 

Ligger värdet nära 0 har vi liten correlation dvs litet samband mellan parametrarna

```python
sns.pairplot(df.sample(100,random_state=42), hue="Quality", corner=True, height=1.0);
```
